### 5. PT: Crie um gráfico de dispersão (scatter) 2D com [componente principal 1] e [componente principal 2] adicionando cor verde para Benigno e vermelho para a coluna de tipos de câncer maligno: rótulo


* Vamos a trazar la visualización de las 569 muestras a lo largo de los ejes del componente principal - 1 y del componente principal - 2. Esto le dará una buena idea de cómo se distribuyen las muestras entre las dos clases

```python

principalDf = pd.DataFrame(data = principalComponents
             , columns = ['principal component 1', 'principal component 2'])
principalDf
```

```python
df['label'].value_counts()
```

```python
principalDf['label']= df['label']
```

```python
finalDf = principalDf.copy()
```

```python
%matplotlib inline
from matplotlib import pyplot as plt
fig = plt.figure(figsize = (8,8))
ax = fig.add_subplot(1,1,1) 
ax.set_xlabel('Principal Component 1', fontsize = 15)
ax.set_ylabel('Principal Component 2', fontsize = 15)
ax.set_title('2 component PCA', fontsize = 20)

targets = ['Malignant', 'Benign']
colors = ['r','b']
for target, color in zip(targets,colors):
    indicesToKeep = finalDf['label'] == target
    ax.scatter(finalDf.loc[indicesToKeep, 'principal component 1']
               , finalDf.loc[indicesToKeep, 'principal component 2']
               , c = color
               , s = 50)
ax.legend(targets)
ax.grid()
```
